```python
import torch
import torch.nn.functional as F
from torch import amp

# -----------------------------
# CONFIG
# -----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

num_epochs = 50
patch_size = 128  # Reduced patch size to alleviate OOM
scale = 4

scaler = amp.GradScaler('cuda')

# -----------------------------
# PATCH EXTRACTION
# -----------------------------
def extract_patch(lr, hr):
    # lr and hr are expected to be (Batch, Channel, Height, Width)
    # The shape should be (B, C, H, W) after `unsqueeze(1)` operation in the loop.
    _, _, h, w = lr.shape  # Correctly unpacks 4 dimensions: Batch, Channel, Height, Width

    x = torch.randint(0, w - patch_size, (1,)).item()
    y = torch.randint(0, h - patch_size, (1,)).item()

    # Slice across batch and channel dimensions
    lr_patch = lr[:, :, y:y+patch_size, x:x+patch_size]
    hr_patch = hr[:, :, y*scale:(y+patch_size)*scale, x*scale:(x+patch_size)*scale]

    return lr_patch, hr_patch


# -----------------------------
# MAIN TRAIN LOOP
# -----------------------------
for epoch in range(num_epochs):
    print(f"Epoch {epoch+1}/{num_epochs}")

    for lr, hr in loader:

        # -----------------------------
        # MOVE TO GPU
        # -----------------------------
        lr = lr.float().to(device)
        hr = hr.float().to(device)

        # -----------------------------
        # FORCE CORRECT SHAPE: (B,C,H,W)
        # Assuming lr and hr from loader are (Batch, Height, Width) for single channel images
        # -----------------------------
        lr = lr.unsqueeze(1) # Add channel dimension (B, H, W) -> (B, 1, H, W)
        hr = hr.unsqueeze(1) # Add channel dimension (B, H_HR, W_HR) -> (B, 1, H_HR, W_HR)

        # -----------------------------
        # RANDOM PATCH (CRITICAL FOR OOM FIX)
        # -----------------------------
        lr, hr = extract_patch(lr, hr)

        optimizer.zero_grad(set_to_none=True)

        # -----------------------------
        # FORWARD PASS (MODERN AMP)
        # Using `amp.autocast('cuda')` from `torch.amp` now correctly specifies device type
        # -----------------------------
        with amp.autocast('cuda'):
            sr = model(lr)
            loss = criterion(sr, hr)

        # -----------------------------
        # BACKPROP
        # -----------------------------
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

    torch.cuda.empty_cache()
```

```python
import os
import torch
import numpy as np

def load_npy_folder(folder):
    imgs = []
    if not os.path.exists(folder):
        raise FileNotFoundError(f"Critical Error: The directory '{folder}' does not exist. Check your paths!")

    for f in sorted(os.listdir(folder)):
        if f.endswith(".npy"):
            path = os.path.join(folder, f)
            img_np = np.load(path)
            img_tensor = torch.tensor(img_np, dtype=torch.float32)

            # Ensures tensor has a channel dimension: (Channels, Height, Width)
            if img_tensor.dim() == 2:
                img_tensor = img_tensor.unsqueeze(0)
            imgs.append(img_tensor)

    return imgs

# Update these strings if your folders are located elsewhere (e.g., "/content/train/lr")
lr_imgs = load_npy_folder("train/lr")
hr_imgs = load_npy_folder("train/hr")

print(f"lr_imgs populated with {len(lr_imgs)} patches.")
print(f"hr_imgs populated with {len(hr_imgs)} patches.")
```

```python
from torch.utils.data import Dataset, DataLoader
import random

class SRPatchDataset(Dataset):
    def __init__(self, lr_images, hr_images, lr_patch_size, scale):
        self.lr_images = lr_images  # List of (C, H, W) tensors
        self.hr_images = hr_images  # List of (C, H_HR, W_HR) tensors
        self.lr_patch_size = lr_patch_size
        self.scale = scale
        self.hr_patch_size = lr_patch_size * scale

    def __len__(self):
        # We need to make sure that lr_images is not empty to avoid error.
        # If the list is empty, something is wrong with data loading.
        if not self.lr_images:
            return 0
        return len(self.lr_images)

    def __getitem__(self, idx):
        lr_img = self.lr_images[idx] # (C, H_orig_LR, W_orig_LR)
        hr_img = self.hr_images[idx] # (C, H_orig_HR, W_orig_HR)

        # Get the spatial dimensions of the original LR image (after channel dim)
        _, h_lr_orig, w_lr_orig = lr_img.shape

        # Ensure there's enough space to extract a patch of lr_patch_size
        if w_lr_orig < self.lr_patch_size or h_lr_orig < self.lr_patch_size:
            raise ValueError(f"Image at index {idx} (LR: {lr_img.shape}, HR: {hr_img.shape}) is too small to extract a patch of size {self.lr_patch_size}")

        x_lr = random.randint(0, w_lr_orig - self.lr_patch_size)
        y_lr = random.randint(0, h_lr_orig - self.lr_patch_size)

        # Extract LR patch
        lr_patch = lr_img[:, y_lr : y_lr + self.lr_patch_size,
                          x_lr : x_lr + self.lr_patch_size] # (C, lr_patch_size, lr_patch_size)

        # Calculate corresponding top-left corner for HR patch
        x_hr = x_lr * self.scale
        y_hr = y_lr * self.scale

        # Extract HR patch
        hr_patch = hr_img[:, y_hr : y_hr + self.hr_patch_size,
                          x_hr : x_hr + self.hr_patch_size] # (C, hr_patch_size, hr_patch_size)

        return lr_patch, hr_patch

# Instantiate the new SRPatchDataset
# `lr_imgs` and `hr_imgs` are assumed to be loaded and available from cell ObTklcKYF09H.

# Using a patch size of 128x128 for LR, which results in 256x256 for HR with a scale of 2.
# This is a good balance for GPU memory and model learning capacity.

train_dataset = SRPatchDataset(lr_imgs, hr_imgs, lr_patch_size=128, scale=2)

# Setup the DataLoader with the new dataset and batch_size=1
train_loader = DataLoader(
    train_dataset,
    batch_size=1,  # Keeping batch size 1 to avoid OOM for now
    shuffle=True,
    num_workers=2, # Keep num_workers = 2 for better performance
    pin_memory=True # Pin memory to potentially speed up data transfer to GPU
)

print(f"SRPatchDataset initialized with {len(train_dataset)} samples.")
print(f"DataLoader created with batch_size={train_loader.batch_size}.")
```

```python
import torch
import torch.nn as nn

class SimpleSRCNN(nn.Module):
    def __init__(self, scale_factor=2):
        super(SimpleSRCNN, self).__init__()
        self.scale_factor = scale_factor
        # Layer 1: Patch extraction and feature representation
        # Assumes 1 channel (grayscale/raster). in_channels=1.
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=64, kernel_size=9, padding=4)
        self.relu1 = nn.ReLU()

        # Layer 2: Non-linear mapping
        self.conv2 = nn.Conv2d(64, 32, kernel_size=5, padding=2)
        self.relu2 = nn.ReLU()

        # Layer 3: Upsampling and Reconstruction
        # Using PixelShuffle for upsampling by scale_factor
        self.upsample = nn.Sequential(
            nn.Conv2d(32, 1 * (scale_factor ** 2), kernel_size=3, padding=1),
            nn.PixelShuffle(scale_factor)
        )

    def forward(self, x):
        x = self.relu1(self.conv1(x))
        x = self.relu2(self.conv2(x))
        x = self.upsample(x)
        return x

# Move the model to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = SimpleSRCNN(scale_factor=2).to(device) # Initialize with scale_factor=2
print(f"Model is initialized and running on: {device}")
```

```python
import torch.optim as optim

# Loss function: Mean Squared Error (pixel-by-pixel comparison)
criterion = nn.MSELoss()

# Optimizer: Adam is fast and robust for this
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
epochs = 20  # You can increase this later (e.g., 50 or 100) for better quality

print("Starting training...")
for epoch in range(epochs):
    running_loss = 0.0

    for lr_batch, hr_batch in train_loader:
        # Move data to GPU if available
        lr_batch = lr_batch.to(device)
        hr_batch = hr_batch.to(device)

        # 1. Clear previous gradients
        optimizer.zero_grad()

        # 2. Forward pass: Predict HR images from LR inputs
        outputs = model(lr_batch)

        # 3. Calculate loss
        loss = criterion(outputs, hr_batch)

        # 4. Backward pass: Calculate gradients
        loss.backward()

        # 5. Optimization step: Update weights
        optimizer.step()

        running_loss += loss.item() * lr_batch.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)
    print(f"Epoch [{epoch+1}/{epochs}], Loss: {epoch_loss:.6f}")

print("Training finished!")
```
